## A priori

```python
atp_ranking = pd.read_csv('web_data/atp_initial_nofiltered_ranking.csv')
```

```python
atp_ranking['player_id'].value_counts()
```

```python
import numpy as np

# Map player elo ratings to a dictionary for fast lookup
elo_dict = dict(zip(atp_ranking["player_id"], atp_ranking["elo_rating"]))

# Convert player IDs to NumPy arrays
winner_ids = atp_matches["winner_id"].values
loser_ids = atp_matches["loser_id"].values

# Lookup Elo ratings using NumPy's fast indexing
winner_elos = np.array([elo_dict[w_id] for w_id in winner_ids])
loser_elos = np.array([elo_dict[l_id] for l_id in loser_ids])

# Compute accuracy using a NumPy operation
correct = np.sum(winner_elos > loser_elos)
accuracy = correct / len(atp_matches)
accuracy
```
